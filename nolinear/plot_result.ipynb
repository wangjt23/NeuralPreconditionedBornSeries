```python
import torch
import matplotlib.pyplot as plt

u_init = torch.load('/afs/jiajiwei-folder/wangjt/work/precond_loss/nolinear/u_initial_fno.pt', weights_only=False)
du = torch.load('/afs/jiajiwei-folder/wangjt/work/precond_loss/nolinear/du_fno.pt', weights_only=False)
iter_his = torch.load('/afs/jiajiwei-folder/wangjt/work/precond_loss/nolinear/iter_his_fno.pt', weights_only=False)
iter_his_pre = torch.load('/afs/jiajiwei-folder/wangjt/work/precond_loss/nolinear/iter_his_fno_precond.pt', weights_only=False)
iter_his_pre_l2 = torch.load('/afs/jiajiwei-folder/wangjt/work/precond_loss/nolinear/iter_his_fno_precond_l2.pt', weights_only=False)
if isinstance(u_init, torch.Tensor) and u_init.dim() >= 2:
    u_plot = u_init.squeeze().detach().cpu().numpy()
else:
    u_plot = u_init

if isinstance(du, torch.Tensor) and du.dim() >= 2:
    du_plot = du.squeeze().detach().cpu().numpy()
else:
    du_plot = du

if isinstance(iter_his, torch.Tensor):
    iter_plot = iter_his.detach().cpu().numpy()
else:
    iter_plot = iter_his

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(u_plot)
plt.title(r'initial $u$')
plt.colorbar()

plt.subplot(1, 3, 2)
plt.imshow(du_plot)
plt.title(r'$\delta u$')
plt.colorbar()

plt.subplot(1, 3, 3)
plt.semilogy(range(len(iter_plot)-1), iter_plot[1:], 'o-',label=r'Direct+$\mathcal{R}_{\mathrm{dir}}$')
plt.semilogy(range(len(iter_his_pre_l2)-1), iter_his_pre_l2[1:], 'o-', label=r'NPBS+$\mathcal{R}_{\mathrm{bs}}^{\ell_2}$')
plt.semilogy(range(len(iter_his_pre)-1), iter_his_pre[1:], 'o-', label=r'NPBS+$\mathcal{R}_{\mathrm{bs}}^{R_\eta}$')
plt.legend()
plt.title('Iteration history')
plt.xlabel('Iter')
plt.ylabel('Rel residual')
plt.grid(True)

plt.tight_layout()
plt.show()
```
